```python
import pandas as pd
pd.set_option('display.max_colwidth', 100)

#df- data frame
df = pd.read_csv("sample_text.csv", on_bad_lines='skip')
df.shape
```

```python
df
```

```python
from sentence_transformers import SentenceTransformer
encoder = SentenceTransformer("all-mpnet-base-v2")
vectors = encoder.encode(df.text.tolist())
vectors.shape
```

```python
vectors
```

```python
dim  = vectors.shape[1] # Dimention of the first shape in the above vectors (7, 768)
dim
```

```python
""" FAISS (Facebook AI Similarity Search) is an open-source library developed by Meta for efficient similarity search and clustering of dense vectors, handling datasets from millions to billions of items """
import faiss 
index = faiss.IndexFlatL2(dim)
index
```

```python
index.add(vectors)
```

```python
search_query = "Looking for a place to visit during the holidays"
vec = encoder.encode(search_query)
vec.shape
```

```python
import numpy as np
svec = np.array(vec).reshape(1,-1)  # making/reshaping the above one dimentionaly array into two dimentional array
svec.shape
```

```python
distances, I = index.search(svec, k=2)
I
```
